## 🤖 Interviewer (LLM)

**Q4 score: 1 / 4**

**What you got right:** Your SGD and mini-batch descriptions are correct. A cost function does measure how wrong a model is.

**What was missing or wrong:**
- You didn't name *which* cost function produces these gradients (it's MSE).
- The `* X_train` term wasn't explained.
- A large learning rate doesn't cause overfitting. That's a common mix-up: the learning rate controls **optimization** (how you walk downhill), not model complexity.

**Model answer:**

1. **Cost:** this is MSE, written as J = (1/2m) Σ(ŷ − y)², where ŷ = θ₀ + θ₁x. The ½ is there so the 2 from differentiating cancels out. Taking partial derivatives:
   - ∂J/∂θ₀ = (1/m) Σ(ŷ − y). θ₀ shifts every prediction equally, so its gradient is just the average error.
   - ∂J/∂θ₁ = (1/m) Σ(ŷ − y)·x. By the chain rule, ∂ŷ/∂θ₁ = x. Each error is weighted by its input: points with large x influence the slope more.
   - The update rule is θ ← θ − α·∇J, which steps against the gradient (downhill).

2. **Learning rate α:**
   - **Too large:** steps overshoot the minimum. The loss oscillates or **diverges**, growing until you get `inf` or `nan`.
   - **Too small:** the loss falls steadily but very slowly, and it may still be far from the minimum when the epochs run out (underfit parameters).
   - **How to spot it:** plot loss against epoch. A healthy curve drops quickly, then flattens. You can also compare θ against `LinearRegression().coef_`, as the workshop did. Unscaled features make all of this worse, which is another reason to scale before using gradient descent.

3. **Variants:**
   - **Batch:** uses all m samples per update. Gradients are smooth and exact, but each step is expensive.
   - **SGD:** uses 1 sample per update. It's fast and noisy, and the noise can help it escape shallow regions. It needs a decaying learning rate to settle. This is what `SGDRegressor` does.
   - **Mini-batch:** uses roughly 32–256 samples per update. It's the practical middle ground: vectorized and fairly stable. It's the default in deep learning.
   - One full pass through the data is an **epoch**. `max_iter` and `tol` (and early stopping) control when training stops.

## 🤖 Interviewer (LLM)

**Q6 score: 0 / 4**

This is a core interview concept, and it ties together several things you already know. Here's a mental model to hold on to: **complexity is a dial**. Turn it too low and the model is too simple to capture the pattern. Turn it too high and it starts memorizing noise.

**Model answer:**

1. **Training vs. validation error as the degree goes up:**
   - **Training error** always goes down (or stays the same). A higher-degree polynomial can always fit the training points at least as well.
   - **Validation error** traces a **U-shape**. It falls at first, then rises once the model starts fitting noise.
   - **Degree 0** predicts a flat line at the mean, so it **underfits**: both errors are high, and R² is about 0.
   - **Degree 1–2** is usually the sweet spot for BMI against diabetes progression, because the relationship is close to linear.
   - **Degree 4–5** starts to **overfit**: training error is a bit lower, validation error is higher, and the curve wiggles and shoots off at the extreme BMI values.
   - **How to choose:** pick the degree with the lowest validation or CV error, not the lowest training error.

2. **Bias-variance trade-off:** expected error = bias² + variance + irreducible noise.
   - **High bias (underfitting):** the model is too rigid and makes systematic errors. Training and validation errors are both high and close together.
   - **High variance (overfitting):** the model is too flexible and changes a lot depending on which training sample it saw. Training error is low and validation error is high, with a big gap between them.
   - Adding complexity lowers bias but raises variance. The goal is the balance point.

3. **The same knob in other models:**
   - **KNN, `n_neighbors` (k):** small k (k = 1) overfits, because the boundary is jagged and follows every noisy point. Large k underfits, because everything gets averaged toward the majority class or the mean.
   - **Decision tree, `max_depth`, `min_samples_split` and `min_samples_leaf`:** a deep tree with tiny leaves overfits (it can memorize the training set, with 1 sample per leaf). A shallow tree with big leaves underfits.
   - **Regularization** (the C parameter in `LogisticRegression`, Ridge/Lasso alpha) is a fourth version of the same knob.

**What you got right:** The core idea. The tree computes MSE for each candidate split and keeps the lowest one.

**What was missing:**

Interviewers love asking for a hand calculation like this.

**Model answer:**

1. **Algorithm, for one feature:**
   - Sort the unique x values.
   - The candidate thresholds are the **midpoints** between neighbouring values: 1.5, 2.5 and 3.5.
   - For each candidate, split the samples into left (x ≤ t) and right (x > t). Compute each child's MSE around **its own mean**.
   - Combine them weighted by size: (n_L·MSE_L + n_R·MSE_R) / n.
   - Choose the threshold with the **lowest weighted MSE**, which is the same as the largest MSE reduction from the parent.
   - With several features, repeat this for every feature and take the best (feature, threshold) pair. Then recurse into each child until a stopping rule is hit (`max_depth`, `min_samples_leaf`, or no further improvement).

2. **The worked example.** The parent mean is 21, and the parent MSE is (121 + 81 + 81 + 121) / 4 = **101**.

   | 1.5 | {10}: 10, 0 | {12, 30, 32}: 24.67, 80.89 | (0 + 3·80.89) / 4 = **60.67** |
   | **2.5** | {10, 12}: 11, 1 | {30, 32}: 31, 1 | (2·1 + 2·1) / 4 = **1.0** ✅ |
   | 3.5 | {10, 12, 30}: 17.33, 80.89 | {32}: 32, 0 | **60.67** |

3. **Classification trees** use an impurity measure instead of MSE:

   - The same weighted-children logic applies, and the threshold that reduces impurity most wins. A leaf predicts the **majority class**, or class proportions for `predict_proba`.

3. **Time series:**
   - Shuffled k-fold lets the model **train on the future and validate on the past**. Neighbouring time steps are highly correlated, so the validation rows are almost copies of training rows. That's leakage, and it makes the scores look far better than live performance would be.

   - A simpler option is a chronological holdout: train on the earliest 80% and test on the latest 20%.
   - Feature engineering must also respect time. Lags and rolling means can only use past values.
